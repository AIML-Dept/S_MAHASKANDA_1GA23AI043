```python
#easy: # Install Qiskit (run once)
!pip install qiskit qiskit-aer

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector

# Create a 2-qubit circuit
circuit = QuantumCircuit(2)

# Set the control qubit q0 to |1>
circuit.x(0)

# Apply CNOT: q0 = control, q1 = target
circuit.cx(0, 1)

# Display circuit
print("CNOT Circuit:")
print(circuit.draw(output="text"))

# Obtain final statevector
final_state = Statevector.from_instruction(circuit)

print("\nFinal Statevector:")
print(final_state)

# Obtain probabilities
prob = final_state.probabilities()

print("\nProbabilities:")
print(prob)

# Verify target qubit has flipped
if prob[3] == 1:
    print("\nVerification: Target qubit flipped successfully.")
    print("Final state is |11>.")
```

```python
#medium:


from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Create a 2-qubit Bell state circuit
bell = QuantumCircuit(2, 2)

# Put first qubit into superposition
bell.h(0)

# Entangle the second qubit with the first
bell.cx(0, 1)

# Measure both qubits
bell.measure_all()

# Display the circuit
print("Bell State Circuit:")
print(bell)

# Simulate the circuit
simulator = AerSimulator()
result = simulator.run(bell, shots=1024).result()

# Get measurement counts
counts = result.get_counts()

print("\nResults after 1024 shots:")
print(counts)

# Verify the Bell-state correlation
if all(state in ["00", "11"] for state in counts):
    print("\nVerification: Only 00 and 11 outcomes were observed.")
else:
    print("\nVerification failed: Unexpected outcome detected.")
```

```python
#hard:


from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

# Create a 3-qubit GHZ circuit
ghz = QuantumCircuit(3, 3)

# Create superposition on the first qubit
ghz.h(0)

# Entangle q1 and q2 with the first qubit
ghz.cx(0, 1)
ghz.cx(0, 2)

# Measure all qubits
ghz.measure_all()

# Display the circuit
print("3-Qubit GHZ Circuit:")
print(ghz)

# Run the circuit for 1024 shots
simulator = AerSimulator()
result = simulator.run(ghz, shots=1024).result()

# Get measurement results
counts = result.get_counts()

print("\nMeasurement Results:")
print(counts)

# Verify three-way correlation
unexpected = set(counts) - {"000", "111"}

print("\nVerification:")

if not unexpected:
    print("Perfect three-way correlation confirmed.")
    print("Only 000 and 111 outcomes were observed.")
else:
    print("Unexpected outcomes:", unexpected)
```

```python
#real world:


from qiskit import QuantumCircuit
from qiskit_aer import AerSimulator

simulator = AerSimulator()

# Full Adder:
# Sum = A XOR B XOR Cin
# Carry = AB XOR Cin(A XOR B)

def full_adder(A, B, Cin):

    # q0 = A, q1 = B, q2 = Cin
    # q3 = Sum, q4 = Carry
    circuit = QuantumCircuit(5, 2)

    # Set input qubits
    if A:
        circuit.x(0)
    if B:
        circuit.x(1)
    if Cin:
        circuit.x(2)

    # Calculate Sum
    circuit.cx(0, 3)
    circuit.cx(1, 3)
    circuit.cx(2, 3)

    # Calculate Carry
    circuit.ccx(0, 1, 4)

    circuit.cx(0, 1)
    circuit.ccx(1, 2, 4)
    circuit.cx(0, 1)

    # Measure Sum and Carry
    circuit.measure(3, 0)
    circuit.measure(4, 1)

    return circuit


print("CLASSICAL FULL ADDER USING QUANTUM GATES")
print("=" * 55)

# Test all possible input combinations
for A in range(2):
    for B in range(2):
        for Cin in range(2):

            circuit = full_adder(A, B, Cin)

            result = simulator.run(circuit, shots=1).result()
            output = next(iter(result.get_counts()))

            # Measurement order: Carry, Sum
            carry = output[0]
            sum_bit = output[1]

            print(
                f"A={A}, B={B}, Cin={Cin} "
                f"--> Sum={sum_bit}, Carry={carry}"
            )

# Display the circuit for one example
print("\nExample Circuit (A=1, B=1, Cin=1):")
print(full_adder(1, 1, 1))
```

```python
#challenge:

from qiskit import QuantumCircuit
from qiskit.quantum_info import Statevector
import numpy as np

# Create a 2-qubit circuit
circuit = QuantumCircuit(2)

# Rotation angle
angle = np.pi / 3

# Create the required amplitudes
circuit.ry(angle, 0)

# Create entanglement between the two qubits
circuit.cx(0, 1)

# Display the circuit
print("Arbitrary 2-Qubit Entangled State Circuit:")
print(circuit)

# Statevector inspection (state tomography)
final_state = Statevector.from_instruction(circuit)

print("\nStatevector:")
print(final_state)

# Calculate probabilities
probs = final_state.probabilities()

print("\nState Probabilities:")
for i, p in enumerate(probs):
    print(f"|{i:02b}> : {p:.2f}")

# Expected state
expected = Statevector([
    np.sqrt(3) / 2,
    0,
    0,
    1 / 2
])

# Verify
print("\nVerification:")

if final_state.equiv(expected):
    print("The generated state matches the chosen arbitrary entangled state.")
    print("It is different from the standard Bell states.")
else:
    print("State verification failed.")
```
